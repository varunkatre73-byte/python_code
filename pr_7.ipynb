```python
import numpy as np
```

```python
# Create a 1-D Array
arr1 = np.array([10, 20, 30, 40, 50])
print("1-D Array:", arr1)
```

```python
# Create a 2-D Array
arr2 = np.array([[1, 2, 3],
 [4, 5, 6]])
print("2-D Array:\n", arr2)
```

```python
# Create Special Arrays
a = np.array([5, 10, 15])
b = np.array([2, 4, 6])
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
# Statistical Functions
data = np.array([12, 15, 18, 20, 25])
print("Mean:", data.mean())
print("Sum:", data.sum())
print("Maximum:", data.max())
print("Minimum:", data.min())
```

```python
# Square Root and Power
arr = np.array([4, 9, 16, 25])
print("Square Root:", np.sqrt(arr))
print("Power (Square):", np.power(arr, 2))
```

```python
# Absolute and Round Functions
arr = np.array([-5.7, 3.2, -8.9, 4.5])
print("Absolute Values:", np.abs(arr))
print("Rounded Values:", np.round(arr))
```
